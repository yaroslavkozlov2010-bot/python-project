# Напиши ниже код, который считает кол-во чисел кратное 3. программа заканчивается на ноль

> **Важное правило:** Никогда не изменяйте переменные в прошлых ячейках, если от них зависит выполнение последующих, чтобы избежать путаницы в памяти блокнота.

````python
count = 0

x = int(input())
while x != 0:
    if x % 3 == 0:
        count += 1
    x = int(input())
````
