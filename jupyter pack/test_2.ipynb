Решим уравнение с помощью библиотеки sympy

Сначал объявим математическую переменную

```python
from sympy import *

x = symbols('x')
```

Теперь заапишем уравнение (я запишу в переменную, но можно обойтись и без нее)

```python
eq = Eq(5*x**3, 43)
```

Теперь скажем программе решить это уравнение относительно нашей переменной и выведем решение

```python
answer = solve(eq, x)
init_printing() # Включает красивый вывод (LaTeX)
display(answer)

end = input()
```
